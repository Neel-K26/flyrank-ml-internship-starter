## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load the starter dataset (note the ../../ because you are in work/notebooks/)
df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

# Number 1: Total pages to evaluate
total_pages = len(df)

# Number 2: The current decline rate in the sample
declining_pages = (df["trend_direction"].str.lower() == "down").sum()
decline_rate = declining_pages / total_pages

# Number 3: How many pages meet a basic "stale" rule
stale_visible = ((df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)).sum()

print(f"Total pages in dataset: {total_pages:,}")
print(f"Pages currently trending down: {declining_pages:,} ({decline_rate:.1%})")
print(f"Pages that are stale (>=180 days) AND visible (>=500 impressions): {stale_visible:,}")
```
