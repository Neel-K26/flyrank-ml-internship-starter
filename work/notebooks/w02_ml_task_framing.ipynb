## 5. Why ML beats a fixed rule here
Why ML wins: A fixed heuristic (like days_since_last_update > 180 AND impressions_90d > 500) is brittle. It treats a 179-day-old page as perfectly healthy and a 181-day-old page as broken. ML replaces these hard cutoffs with a continuous risk probability by weighing non-linear interactions—for instance, realizing that a young page with rapidly dropping CTR might actually need a refresh much sooner than an old page with perfectly stable engagement.
